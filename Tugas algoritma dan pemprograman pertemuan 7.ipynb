```
while True:
    print("\n=== MENU ===")
    print("1. A pangkat B")
    print("2. Deret: 1 - 2/3 + 5/8 - 13/21")
    print("0. Keluar")

    pilih = input("Pilih menu: ")

    if pilih == "1":
        try:
            angka = int(input("Masukkan bilangan: "))
            pangkat = int(input("Masukkan nilai pangkat: "))

            i = 1
            while i <= pangkat:
                print(f"{angka}^{i} = {angka ** i}")
                i += 1

        except ValueError:
            print("Input harus berupa bilangan bulat!")

    elif pilih == "2":
        try:
            n = int(input("Masukkan jumlah suku: "))
            total = 0

            # generate fibonacci
            fib = [1, 1]
            for _ in range(n * 2 - 2):
                fib.append(fib[-1] + fib[-2])

            print("Deret:", end=" ")
            for i in range(n):
                atas = fib[2 * i]
                bawah = fib[2 * i + 1]
                nilai = atas / bawah

                if i % 2 == 0:
                    total += nilai
                    tanda = "+" if i != 0 else ""
                else:
                    total -= nilai
                    tanda = "-"

                print(f"{tanda}({atas}/{bawah})", end=" ")

            print(f"\nTotal = {total}")

        except ValueError:
            print("Input harus angka!")

    elif pilih == "0":
        print("Keluar dari program...")
        break

    else:
        print("Menu tidak tersedia!")
```
