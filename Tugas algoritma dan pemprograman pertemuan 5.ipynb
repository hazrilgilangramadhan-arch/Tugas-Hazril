```
nama = ["Asep", "Budi", "Cecep"]

nilai = [
    [50, 70, 40, 80],
    [78, 78, 80, 65],
    [57, 88, 67, 69]
]

# Rata-rata mahasiswa
rata_mhs = []
for i in range(3):
    total = sum(nilai[i])
    rata_mhs.append(total / 4)

# Cari mahasiswa terbaik
idx_max = 0
for i in range(1, 3):
    if rata_mhs[i] > rata_mhs[idx_max]:
        idx_max = i

# Rata-rata mata kuliah
rata_mk = []
for j in range(4):
    total = 0
    for i in range(3):
        total += nilai[i][j]
    rata_mk.append(total / 3)

# Cari mata kuliah terkecil
idx_min = 0
for j in range(1, 4):
    if rata_mk[j] < rata_mk[idx_min]:
        idx_min = j

# Output
print(f"Mahasiswa Terpintar : {nama[idx_max]} (Nilai : {rata_mhs[idx_max]:.2f})")
print(f"Mata Kuliah Nilai Terkecil : MK{idx_min+1} (Nilai : {rata_mk[idx_min]:.2f})")
```

```
# fungsi input matriks
def input_matriks(baris, kolom):
    matriks = []
    for i in range(baris):
        row = []
        for j in range(kolom):
            nilai = int(input(f"Masukkan nilai [{i}][{j}]: "))
            row.append(nilai)
        matriks.append(row)
    return matriks


# fungsi tampil matriks
def tampil(m):
    for row in m:
        print(row)


# penjumlahan
def tambah(A, B):
    hasil = []
    for i in range(len(A)):
        row = []
        for j in range(len(A[0])):
            row.append(A[i][j] + B[i][j])
        hasil.append(row)
    return hasil


# pengurangan
def kurang(A, B):
    hasil = []
    for i in range(len(A)):
        row = []
        for j in range(len(A[0])):
            row.append(A[i][j] - B[i][j])
        hasil.append(row)
    return hasil


# perkalian
def kali(A, B):
    hasil = []
    for i in range(len(A)):
        row = []
        for j in range(len(B[0])):
            total = 0
            for k in range(len(B)):
                total += A[i][k] * B[k][j]
            row.append(total)
        hasil.append(row)
    return hasil


# MENU
while True:
    print("\n=== MENU MATRIKS ===")
    print("1. Penjumlahan")
    print("2. Pengurangan")
    print("3. Perkalian")
    print("0. Exit")

    pilih = input("Pilih menu: ")

    if pilih == "1" or pilih == "2":
        baris = int(input("Masukkan jumlah baris: "))
        kolom = int(input("Masukkan jumlah kolom: "))

        print("\nMatriks A")
        A = input_matriks(baris, kolom)

        print("\nMatriks B")
        B = input_matriks(baris, kolom)

        if pilih == "1":
            hasil = tambah(A, B)
            print("\nHasil Penjumlahan:")
        else:
            hasil = kurang(A, B)
            print("\nHasil Pengurangan:")

        tampil(hasil)

    elif pilih == "3":
        baris_A = int(input("Baris A: "))
        kolom_A = int(input("Kolom A: "))
        baris_B = int(input("Baris B: "))
        kolom_B = int(input("Kolom B: "))

        if kolom_A != baris_B:
            print("Tidak bisa dikalikan! (kolom A harus = baris B)")
        else:
            print("\nMatriks A")
            A = input_matriks(baris_A, kolom_A)

            print("\nMatriks B")
            B = input_matriks(baris_B, kolom_B)

            hasil = kali(A, B)
            print("\nHasil Perkalian:")
            tampil(hasil)

    elif pilih == "0":
        print("Program selesai.")
        break

    else:
        print("Pilihan tidak valid!")
```
